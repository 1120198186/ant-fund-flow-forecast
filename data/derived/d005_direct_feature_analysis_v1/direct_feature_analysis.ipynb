# Direct feature and target associations

Reviewed evidence at three grains: user demographics, daily rates, and daily/user transaction components.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display
root = Path.cwd() / 'data' / 'derived' / 'd005_direct_feature_analysis_v1'
demo = pd.read_csv(root / 'demographic_associations.csv')
rates = pd.read_csv(root / 'rate_associations.csv')
transactions = pd.read_csv(root / 'transaction_component_associations.csv')
catalog = pd.read_csv(root / 'feature_combination_catalog.csv')
```

```python
primary = ['total_purchase_amt','total_redeem_amt','direct_purchase_amt','purchase_bank_amt','transfer_amt']
display(demo[(demo.target.isin(primary)) & (demo.metric == 'log1p_amount_per_exposure_day')].sort_values('eta_squared', ascending=False))
```

```python
cols = ['feature','target','spearman','diff1_corr_log_target','lagged_diff1_corr_log_target']
display(rates[rates.target.isin(primary)].sort_values('short_run_abs_diff1', ascending=False)[cols].head(25))
```
